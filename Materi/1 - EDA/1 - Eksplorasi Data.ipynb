Eksplorasi data adalah tahap pertama dalam proses data mining. Tujuannya bukan membangun model, melainkan memahami data terlebih dahulu: apa saja atributnya, bagaimana bentuk sebarannya, dan hubungan apa yang muncul di antara atribut-atributnya.

### 2.1.1. Atribut Nominal (Nominal attributes)

#### Atribut Nominal:

Atribut nominal adalah variabel kategorik yang menyatakan kategori atau label berbeda tanpa urutan maupun peringkat di antaranya. Atribut ini mengelompokkan data berdasarkan perbedaan kualitatif. Contohnya warna, jenis kelamin, jenis hewan, dan sejenisnya.

Pada dataset yang dipakai mulai Section 2.2, `kota` dan `metode_bayar` adalah contoh atribut nominal: nilai `Jakarta` dan `Bandung` hanya berbeda, tidak ada yang lebih tinggi. `id_pelanggan` juga nominal, tetapi nilainya unik untuk hampir setiap baris sehingga berperan sebagai pengenal, bukan fitur.

#### Kode Python untuk Menangani Atribut Nominal:

Berikut potongan kode Python sederhana menggunakan library pandas untuk mengolah dataset yang memuat atribut nominal. Dataset contoh dimuat, atribut nominalnya dieksplorasi, lalu dilakukan one-hot encoding.

```python
import pandas as pd

# Sample dataset with nominal attributes
data = {
    'Animal': ['Dog', 'Cat', 'Fish', 'Bird', 'Snake'],
    'Color': ['Brown', 'Black', 'Gold', 'Blue', 'Green'],
    'Habitat': ['Forest', 'Home', 'Aquarium', 'Sky', 'Jungle']
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
print("\n")

# Explore the nominal attributes
nominal_attributes = ['Animal', 'Color', 'Habitat']

# Display the unique values in each nominal attribute
for attribute in nominal_attributes:
    print(f"Unique values in {attribute}: {df[attribute].unique()}")

# Perform one-hot encoding
df_encoded = pd.get_dummies(df, columns=nominal_attributes)

# Display the dataset after one-hot encoding
print("\nDataset after One-Hot Encoding:")
print(df_encoded)
```

In this Example:

    Sebuah pandas DataFrame dibuat dengan tiga atribut nominal: 'Animal', 'Color', dan 'Habitat'.
    Nilai unik pada setiap atribut nominal dieksplorasi.
    Fungsi pd.get_dummies() dipakai untuk melakukan one-hot encoding, yang mengubah setiap atribut nominal menjadi kolom-kolom biner.
    DataFrame hasilnya (df_encoded) ditampilkan setelah one-hot encoding.

### 2.1.2. Atribut Biner (Binary attributes)

#### Atribut Biner:

Atribut biner adalah variabel kategorik yang hanya dapat bernilai salah satu dari dua kemungkinan, umumnya menyatakan ada atau tidak adanya suatu karakteristik. Atribut ini mendasar pada banyak dataset, contohnya Yes/No, True/False, 1/0, atau dua kategori berbeda lainnya.

Pada dataset Section 2.2, `jenis_kelamin` (L/P) adalah biner simetris karena kedua nilainya sama pentingnya. `pernah_komplain` (Ya/Tidak) adalah biner asimetris: nilai `Ya` hanya muncul pada 120 dari 1.503 baris, dan justru kemunculan yang jarang itulah yang menarik untuk dianalisis. `berlangganan_premium` juga bersifat biner.

#### Kode Python untuk Menangani Atribut Biner:

Berikut potongan kode Python menggunakan library pandas untuk mengolah dataset yang memuat atribut biner. Dataset contoh dimuat, atribut binernya dieksplorasi, lalu dilakukan operasi dasar terhadapnya.

```python
import pandas as pd

# Sample dataset with binary attributes
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'PassedExam': [1, 0, 1, 1, 0],
    'EnrolledInCourse': [1, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
print("\n")

# Explore the binary attributes
binary_attributes = ['PassedExam', 'EnrolledInCourse']

# Display the count of each unique value in binary attributes
for attribute in binary_attributes:
    print(f"Counts for {attribute}:\n{df[attribute].value_counts()}\n")

# Perform basic operations on binary attributes
df['TotalAttributes'] = df['PassedExam'] + df['EnrolledInCourse']

# Display the dataset after the operation
print("Dataset after performing an operation on binary attributes:")
print(df)
```

In this Example:

    Sebuah pandas DataFrame dibuat dengan dua atribut biner: 'PassedExam' dan 'EnrolledInCourse'.
    Jumlah kemunculan setiap nilai unik pada atribut biner dieksplorasi menggunakan value_counts().
    Operasi dasar berupa penjumlahan dilakukan terhadap atribut biner untuk membentuk atribut baru, 'TotalAttributes'.
    DataFrame hasilnya (df) ditampilkan setelah operasi tersebut.

### 2.1.3. Atribut Ordinal (Ordinal attributes)

#### Atribut Ordinal:

Atribut ordinal adalah variabel kategorik yang memiliki urutan atau peringkat yang bermakna di antara kategorinya. Berbeda dengan atribut nominal, atribut ordinal punya urutan yang jelas dan bermakna, tetapi jarak antar nilainya tidak harus seragam atau terdefinisi dengan baik. Contoh atribut ordinal adalah jenjang pendidikan (misalnya SD, SMA, perguruan tinggi), tingkat kepuasan pelanggan, atau kelas sosial ekonomi.

Pada dataset Section 2.2 ada dua atribut ordinal. `kepuasan` punya lima tingkat: Sangat Rendah < Rendah < Sedang < Tinggi < Sangat Tinggi. `tingkat_membership` punya tiga tingkat: Bronze < Silver < Gold. Urutan ini tidak dikenali pandas secara otomatis - tanpa diberi tahu, pandas mengurutkannya menurut abjad sehingga `Sangat Tinggi` muncul sebelum `Sedang`. Cara memberitahukannya ditunjukkan pada 2.2.0.

#### Kode Python untuk Menangani Atribut Ordinal:

Berikut potongan kode Python menggunakan library pandas untuk mengolah dataset yang memuat atribut ordinal. Dataset contoh dimuat, atribut ordinalnya dieksplorasi, lalu diperlihatkan cara melakukan encoding yang mempertahankan hubungan ordinalnya.

```python
import pandas as pd

# Sample dataset with ordinal attributes
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'EducationLevel': ['High School', 'College', 'Elementary', 'College', 'High School'],
    'SatisfactionRating': [3, 5, 2, 4, 1]
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
print("\n")

# Explore the ordinal attributes
ordinal_attributes = ['EducationLevel', 'SatisfactionRating']

# Display the unique values in each ordinal attribute
for attribute in ordinal_attributes:
    print(f"Unique values in {attribute}: {df[attribute].unique()}")

# Encode ordinal attributes with meaningful numerical values
education_level_mapping = {'Elementary': 1, 'High School': 2, 'College': 3}
df['EducationLevelEncoded'] = df['EducationLevel'].map(education_level_mapping)

# Display the dataset after encoding ordinal attributes
print("\nDataset after encoding ordinal attributes:")
print(df)
```

In this Example:

    Sebuah pandas DataFrame dibuat dengan dua atribut ordinal: 'EducationLevel' dan 'SatisfactionRating'.
    Nilai unik pada setiap atribut ordinal dieksplorasi.
    Atribut ordinal di-encode menjadi nilai numerik yang bermakna menggunakan fungsi map() untuk membentuk atribut baru, 'EducationLevelEncoded'.
    DataFrame hasilnya (df) ditampilkan setelah operasi tersebut.

### 2.1.4. Atribut Numerik (Numeric attributes)

#### Atribut Numerik:

Atribut numerik menyatakan kuantitas dan bernilai angka. Ada dua jenis utama atribut numerik: diskrit dan kontinu. Atribut numerik diskrit hanya dapat bernilai angka yang terpisah satu sama lain (misalnya jumlah kamar tidur dalam sebuah rumah), sedangkan atribut numerik kontinu dapat bernilai apa saja dalam suatu rentang (misalnya tinggi dan berat badan).

Pada dataset Section 2.2, `total_belanja` dan `usia` bersifat rasio karena nol pada keduanya berarti benar-benar tidak ada, sehingga pernyataan "dua kali lipat" bermakna. Sebaliknya `tahun_bergabung` bersifat interval: selisih antara 2020 dan 2022 bermakna dua tahun, tetapi tahun 2022 bukan berarti dua kali lipat tahun 1011. Karena itu rata-rata `tahun_bergabung` yang muncul pada 2.2.1 perlu dibaca dengan hati-hati.

#### Kode Python untuk Menangani Atribut Numerik:

Berikut potongan kode Python menggunakan library pandas untuk mengolah dataset yang memuat atribut numerik. Dataset contoh dimuat, atribut numeriknya dieksplorasi, lalu dilakukan operasi dasar terhadapnya.

```python
import pandas as pd

# Sample dataset with numeric attributes
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'Age': [21, 19, 22, 20, 23],
    'Height (cm)': [175, 160, 180, 165, 185],
    'Score': [85, 92, 78, 89, 95]
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
print("\n")

# Explore the numeric attributes
numeric_attributes = ['Age', 'Height (cm)', 'Score']

# Display summary statistics for numeric attributes
print("Summary Statistics for Numeric Attributes:")
print(df[numeric_attributes].describe())

# Perform basic operations on numeric attributes
df['NormalizedScore'] = (df['Score'] - df['Score'].mean()) / df['Score'].std()

# Display the dataset after the operation
print("\nDataset after performing an operation on numeric attributes:")
print(df)
```

In this Example:

    Sebuah pandas DataFrame dibuat dengan tiga atribut numerik: 'Age', 'Height (cm)', dan 'Score'.
    Statistik ringkasan untuk atribut numerik dieksplorasi menggunakan describe().
    Operasi dasar berupa normalisasi dilakukan terhadap atribut 'Score' untuk membentuk atribut baru, 'NormalizedScore'.
    DataFrame hasilnya (df) ditampilkan setelah operasi tersebut.

### 2.1.5. Atribut Diskrit dan Kontinu (Discrete vs. continuous attributes)

#### Atribut Diskrit:

Definisi: Atribut diskrit hanya dapat bernilai angka yang terpisah satu sama lain.
Contoh: jumlah kamar tidur dalam rumah, jumlah barang dalam keranjang belanja, jumlah mahasiswa dalam satu kelas.
Sifat: Atribut ini umumnya dihitung dalam bilangan bulat dan memiliki batas yang jelas antar nilainya.

#### Atribut Kontinu:

Definisi: Atribut kontinu dapat bernilai apa saja dalam suatu rentang.
Contoh: tinggi badan, berat badan, suhu, dan segala pengukuran yang dapat bernilai desimal.
Sifat: Atribut ini memiliki himpunan nilai yang kontinu dan tak hingga, sehingga cocok untuk hasil pengukuran.

Pada dataset Section 2.2, `jumlah_transaksi` dan `jumlah_produk_dilihat` bersifat diskrit karena hanya bernilai bilangan bulat hasil pencacahan. `total_belanja`, `lama_langganan_bulan`, dan `rata_rata_rating` bersifat kontinu karena dapat bernilai desimal dalam suatu rentang. Perbedaan ini menentukan pilihan grafik: atribut diskrit dengan sedikit nilai unik lebih cocok digambarkan dengan bar chart, sedangkan atribut kontinu dengan histogram.

#### Kode Python untuk Atribut Diskrit dan Kontinu:

Berikut potongan kode Python menggunakan library pandas untuk mengolah dataset yang memuat atribut diskrit sekaligus kontinu. Dataset contoh dimuat, sifat masing-masing tipe dieksplorasi, lalu dilakukan operasi dasar.

```python
import pandas as pd

# Sample dataset with discrete and continuous attributes
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'NumCourses': [4, 5, 3, 6, 4],  # Discrete attribute (number of courses)
    'GPA': [3.5, 4.0, 3.2, 3.8, 3.9],  # Continuous attribute (GPA)
    'Income': [25000, 30000, 20000, 35000, 32000]  # Continuous attribute (income in dollars)
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
print("\n")

# Explore the nature of attributes
print("Nature of Attributes:")
for column in df.columns:
    if df[column].dtype == 'int':
        print(f"{column} is a discrete attribute.")
    elif df[column].dtype == 'float':
        print(f"{column} is a continuous attribute.")

# Perform basic operations on continuous attributes
df['ScaledIncome'] = df['Income'] / 1000  # Scale income for readability

# Display the dataset after the operation
print("\nDataset after performing an operation on continuous attributes:")
print(df)
```

In this Example:

    Sebuah pandas DataFrame dibuat dengan tiga atribut: 'NumCourses' (diskrit), 'GPA' (kontinu), dan 'Income' (kontinu).
    Sifat setiap atribut dieksplorasi berdasarkan tipe datanya.
    Operasi dasar berupa penskalaan dilakukan terhadap atribut kontinu ('Income') untuk membentuk atribut baru, 'ScaledIncome'.
    DataFrame hasilnya (df) ditampilkan setelah operasi tersebut.

#### Nilai tidak valid, tidak konsisten, dan pemeriksaan rentang

Nilai tidak valid mustahil menurut arti kolomnya. Nilai tidak konsisten adalah beberapa penulisan untuk maksud yang sama. Batas wajar ditetapkan dari pengetahuan tentang domainnya, bukan dari datanya sendiri.

```python
print("usia bernilai negatif   :", df.loc[df["usia"] < 0, "usia"].tolist())
print("rating di luar skala 1-5:", df.loc[df["rata_rata_rating"] > 5, "rata_rata_rating"].tolist())

batas_wajar = {"usia": (0, 100), "rata_rata_rating": (1, 5), "tahun_bergabung": (2020, 2025)}
for kolom, (bawah, atas) in batas_wajar.items():
    di_luar = int(((df[kolom] < bawah) | (df[kolom] > atas)).sum())
    print(f"{kolom:18s} aktual [{df[kolom].min()}, {df[kolom].max()}] -> di luar batas: {di_luar}")

print("\nPenulisan kategori pada kolom kota:")
print(df["kota"].value_counts())
```

#### Pembersihan minimal

Yang dibersihkan hanya yang tidak memerlukan asumsi tambahan: penulisan kota diseragamkan, nilai mustahil jadi `NaN`, duplikat penuh dibuang, dan dua atribut ordinal diberi urutan. Missing value dan outlier sengaja dibiarkan karena itu keputusan modul praproses.

```python
df["kota"] = df["kota"].str.strip().str.title()
df.loc[df["usia"] < 0, "usia"] = np.nan
df.loc[df["rata_rata_rating"] > 5, "rata_rata_rating"] = np.nan
df = df.drop_duplicates().reset_index(drop=True)

urut_kepuasan = ["Sangat Rendah", "Rendah", "Sedang", "Tinggi", "Sangat Tinggi"]
urut_membership = ["Bronze", "Silver", "Gold"]
df["kepuasan"] = pd.Categorical(df["kepuasan"], categories=urut_kepuasan, ordered=True)
df["tingkat_membership"] = pd.Categorical(df["tingkat_membership"],
                                          categories=urut_membership, ordered=True)

print("Ukuran data setelah pembersihan minimal:", df.shape)
print("Kategori kota:", sorted(df["kota"].unique()))
print("Sisa nilai kosong per kolom:")
print(df.isna().sum()[lambda s: s > 0])
```

**Yang dibaca dari output.** Data menyusut 1.503 menjadi 1.500 baris dan `kota` tinggal 6 kategori.

Nilai kosong justru bertambah: `usia` yang tadinya penuh kini punya 2 kosong. Itu memang tujuannya - nilai mustahil kini tercatat "tidak diketahui", bukan angka salah yang ikut terhitung. Tanpa `pd.Categorical`, pandas mengurutkan kepuasan menurut abjad sehingga `Sangat Tinggi` muncul sebelum `Sedang`.

### 2.2.1. Mengukur Pemusatan Data (Measuring the central tendency)

**Mean** peka terhadap nilai ekstrem, **median** tahan terhadapnya, dan **modus** satu-satunya ukuran yang berlaku untuk atribut nominal. Pada sebaran menceng, selisih mean dan median justru menjadi informasi.

```python
pemusatan = pd.DataFrame({
    "mean": df[kolom_numerik].mean(),
    "median": df[kolom_numerik].median(),
    "modus": [df[k].mode().iloc[0] for k in kolom_numerik],
})
pemusatan["selisih_persen"] = ((pemusatan["mean"] - pemusatan["median"])
                               / pemusatan["median"] * 100).round(1)
print(pemusatan.round(2))

print("\nModus atribut kategorik:")
for kolom in ["kota", "metode_bayar", "kepuasan", "tingkat_membership"]:
    print(f"  {kolom:20s}: {df[kolom].mode().iloc[0]}")
```

**Yang dibaca dari output.** Kolom `selisih_persen` menunjukkan kolom mana yang sebarannya tidak simetris. `total_belanja` menonjol sendirian dengan +39,1% (mean 2.245.037 berbanding median 1.614.000), sementara enam kolom lain di bawah 4%.

Modus `total_belanja` 49.000, jauh dari keduanya - modus memang kurang berguna untuk kolom kontinu. Sebaliknya untuk atribut kategorik modus satu-satunya yang masuk akal: menghitung rata-rata dari `Jakarta` dan `Bandung` tidak ada artinya.

### 2.2.2. Mengukur Penyebaran Data (Measuring the dispersion of data)

Dua kelompok bisa punya mean sama persis tetapi sangat berbeda sebarannya. **Range** ditentukan dua nilai ekstrem saja. **Variance** bersatuan kuadrat, **standard deviation** akarnya. **IQR** lebar 50% data tengah, tahan nilai ekstrem. **Coefficient of variation** (std dibagi mean) dipakai membandingkan kolom yang satuannya berbeda.

```python
sebaran = pd.DataFrame({
    "range": df[kolom_numerik].max() - df[kolom_numerik].min(),
    "variance": df[kolom_numerik].var(),
    "std": df[kolom_numerik].std(),
    "IQR": df[kolom_numerik].quantile(.75) - df[kolom_numerik].quantile(.25),
})
sebaran["CV_persen"] = (df[kolom_numerik].std() / df[kolom_numerik].mean() * 100).round(1)
print(sebaran.round(2))

print("\nRingkasan lima angka total_belanja:")
print(df["total_belanja"].describe()[["min", "25%", "50%", "75%", "max"]].round(0))
print("\nQuantile atas total_belanja:")
print(df["total_belanja"].quantile([.90, .95, .99]).round(0))
```

**Yang dibaca dari output.** Variance `total_belanja` 4,66 x 10^12 karena satuannya rupiah kuadrat - tidak bermakna, sehingga yang dilaporkan std-nya (sekitar 2,16 juta).

Std pun tidak bisa dipakai membandingkan antar kolom, karena satuannya berbeda. Untuk itu ada CV: `total_belanja` 96,2% paling menyebar, sedangkan `rata_rata_rating` hanya 15,5% - rating pelanggan cenderung seragam.

Pada lima angka `total_belanja`, jarak minimum ke Q1 hanya 784.000 sedangkan Q3 ke maksimum 14.979.000, sekitar 19 kali lipat. Dari persentil 90 (4.903.800) ke maksimum (17.977.000) melonjak lebih dari 13 juta.

### 2.2.3. Analisis Covariance dan Correlation (Covariance and correlation analysis)

Covariance menunjukkan arah dua variabel bergerak, tetapi besarannya bergantung satuan sehingga tidak bisa dibandingkan antar pasangan. Correlation membaginya dengan hasil kali standard deviation, sehingga nilainya selalu -1 sampai 1. **Pearson** mengukur hubungan garis lurus; **Spearman** menghitung Pearson atas peringkat, sehingga lebih tahan outlier.

```python
pasangan = ["total_belanja", "jumlah_transaksi", "lama_langganan_bulan"]

print("Matriks covariance:")
print(df[pasangan].cov().round(1))
print("\nMatriks correlation (Pearson):")
print(df[pasangan].corr().round(3))
```

**Yang dibaca dari output.** Covariance `total_belanja` ~ `jumlah_transaksi` 13.584.195,5 sedangkan `jumlah_transaksi` ~ `lama_langganan_bulan` hanya 149,5 - seolah pasangan pertama jauh lebih berhubungan. Correlation membalik kesimpulan itu: 0,619 berbanding 0,865. Covariance besar tadi semata karena `total_belanja` bersatuan rupiah jutaan.

*Pertanyaan diskusi:* korelasi `jumlah_transaksi` dengan `lama_langganan_bulan` 0,865. Apakah itu berarti berlangganan lebih lama menyebabkan lebih banyak transaksi?

```python
uji = [("total_belanja", "jumlah_transaksi"),
       ("total_belanja", "lama_langganan_bulan"),
       ("lama_langganan_bulan", "jumlah_transaksi"),
       ("tahun_bergabung", "lama_langganan_bulan"),
       ("jumlah_produk_dilihat", "rata_rata_rating")]

banding = pd.DataFrame([{"pasangan": f"{a} ~ {b}",
                         "pearson": round(df[a].corr(df[b]), 3),
                         "spearman": round(df[a].corr(df[b], method="spearman"), 3)}
                        for a, b in uji])
banding["selisih"] = (banding["spearman"] - banding["pearson"]).round(3)
banding
```

**Yang dibaca dari output.** Dua pasangan pertama melibatkan `total_belanja` dan selisihnya besar (0,619 jadi 0,765; 0,556 jadi 0,681) karena outlier menarik garis lurus Pearson, sementara Spearman hanya memakai peringkat. Selisih besar hampir selalu menandakan outlier atau hubungan melengkung.

`tahun_bergabung` ~ `lama_langganan_bulan` -0,979 bukan temuan, melainkan tanda satu kolom turunan dari kolom lain; memakai keduanya sekaligus menimbulkan multikolinearitas. Pasangan terakhir -0,116 dan -0,005 sama-sama mendekati nol, dan menyimpulkan "tidak ada hubungan" di situ keliru.

#### Deteksi Outlier dengan Aturan IQR

Box plot menandai outlier secara visual tetapi tidak menyebut jumlah dan barisnya. Aturan IQR membuatnya terhitung: nilai di bawah Q1 - 1,5 x IQR atau di atas Q3 + 1,5 x IQR.

**Outlier bukan sinonim kesalahan data.** `usia` -21 jelas kesalahan, tetapi pelanggan yang berbelanja 17,9 juta bisa jadi justru pelanggan terbaik toko ini.

```python
def deteksi_outlier_iqr(seri):
    q1, q3 = seri.quantile(.25), seri.quantile(.75)
    iqr = q3 - q1
    batas_bawah, batas_atas = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    penanda = (seri < batas_bawah) | (seri > batas_atas)
    return batas_bawah, batas_atas, penanda

ringkas_outlier = []
for kolom in kolom_numerik:
    seri = df[kolom].dropna()
    bawah, atas, penanda = deteksi_outlier_iqr(seri)
    ringkas_outlier.append({"kolom": kolom,
                            "batas_bawah": round(bawah, 2), "batas_atas": round(atas, 2),
                            "jumlah_outlier": int(penanda.sum()),
                            "persen": round(penanda.mean() * 100, 2)})
pd.DataFrame(ringkas_outlier)
```

**Yang dibaca dari output.** `total_belanja` punya 72 outlier (5,05%), terbanyak. Batas bawahnya -2.452.000, mustahil untuk nilai belanja - pada sebaran sangat menceng, batas bawah IQR sering jatuh di wilayah tak mungkin sehingga seluruh outlier pasti di sisi kanan.

`rata_rata_rating` punya 34 outlier (2,34%) yang justru semuanya di bawah 2,60, karena kolomnya menceng kiri. Tiga kolom tanpa outlier sama sekali: `jumlah_transaksi`, `lama_langganan_bulan`, `tahun_bergabung`.

```python
seri_belanja = df["total_belanja"].dropna()
_, _, penanda = deteksi_outlier_iqr(seri_belanja)
baris_ekstrem = df.loc[penanda[penanda].index]

print("Jumlah pelanggan dengan total_belanja ekstrem:", len(baris_ekstrem))
print("\nProfil kelompok ekstrem vs seluruh data:")
print(pd.DataFrame({
    "ekstrem": baris_ekstrem[["total_belanja", "jumlah_transaksi", "lama_langganan_bulan"]].mean(),
    "semua": df[["total_belanja", "jumlah_transaksi", "lama_langganan_bulan"]].mean(),
}).round(2))
print("\n5 nilai tertinggi:")
print(baris_ekstrem.nlargest(5, "total_belanja")[
    ["id_pelanggan", "kota", "total_belanja", "jumlah_transaksi", "tingkat_membership"]])
```

**Yang dibaca dari output.** Rata-rata belanja kelompok ekstrem 8.740.639, hampir empat kali rata-rata keseluruhan. Dua kolom lain ikut bergerak: transaksi 28,15 berbanding 16,81, langganan 46,54 berbanding 30,25 bulan. Lima nilai tertinggi seluruhnya Gold.

Jadi 72 baris ini pelanggan setia bernilai tinggi, bukan kesalahan pencatatan - membuangnya berarti membuang segmen paling menguntungkan.

### 2.2.5. Skewness dan Kurtosis

**Skewness** mengukur ketidaksimetrisan: positif berarti ekor memanjang ke kanan, negatif ke kiri. Besaran di bawah 0,5 dianggap hampir simetris, di atas 1 sangat menceng. **Kurtosis** mengukur ketebalan ekor; pandas memakai excess kurtosis sehingga sebaran normal bernilai 0, positif berarti nilai ekstrem lebih sering muncul, negatif berarti sebarannya mendatar.

```python
bentuk = pd.DataFrame({
    "skewness": df[kolom_numerik].skew(),
    "kurtosis": df[kolom_numerik].kurt(),
}).round(2)
bentuk["keterangan"] = np.where(bentuk["skewness"] > .5, "menceng kanan",
                        np.where(bentuk["skewness"] < -.5, "menceng kiri", "hampir simetris"))
bentuk
```

**Yang dibaca dari output.** `total_belanja` berdiri sendiri dengan skewness +2,22 dan kurtosis +7,90, konsisten dengan 72 outlier tadi - kandidat utama transformasi logaritma.

`rata_rata_rating` menceng kiri (-0,92) karena mayoritas memberi rating tinggi sehingga yang jarang justru rating rendah. Lima kolom sisanya hampir simetris, dengan `lama_langganan_bulan` (-1,17) dan `tahun_bergabung` (-1,27) berkurtosis sangat negatif: sebarannya mendatar, bukan lonceng.

```python
fig, ax = plt.subplots(1, 3, figsize=(14, 3.8))
for sumbu, kolom in zip(ax, ["total_belanja", "rata_rata_rating", "usia"]):
    data = df[kolom].dropna()
    sumbu.hist(data, bins=30, color="#4C72B0", edgecolor="white")
    sumbu.axvline(data.mean(), color="red", ls="--", label="mean")
    sumbu.axvline(data.median(), color="black", ls="-", label="median")
    sumbu.set_title(f"{kolom}\nskew = {data.skew():.2f}")
    sumbu.legend()
plt.tight_layout()
plt.show()
```

**Yang dibaca dari output.** Pada `total_belanja` garis mean jelas di kanan median, ditarik ekor panjang. Pada `rata_rata_rating` terbalik. Pada `usia` keduanya nyaris berimpit.

Aturan praktisnya: **pada sebaran menceng, mean tertarik ke arah ekor sementara median bertahan.** Karena itu median lebih layak dilaporkan untuk kolom yang menceng.

### 2.3.1. Data Matrix dan Dissimilarity Matrix (Data matrix vs. dissimilarity matrix)

Dalam data mining, pemahaman tentang konsep data matrix dan dissimilarity matrix sangatlah penting.

#### Data Matrix:

Penjelasan: Data matrix adalah representasi terstruktur dari sebuah dataset, dengan baris yang mewakili masing-masing pengamatan atau instance, dan kolom yang mewakili atribut atau fitur yang berbeda. Setiap sel dalam matriks berisi nilai suatu atribut untuk suatu instance tertentu.

#### Dissimilarity Matrix:

Penjelasan: Dissimilarity matrix merepresentasikan ketidakmiripan atau kemiripan antara pasangan instance dalam sebuah dataset. Matriks ini mengukur seberapa berbeda atau seberapa mirip dua instance berdasarkan suatu ukuran ketidakmiripan, misalnya jarak atau skor dissimilarity.

```python
# Data Matrix

import pandas as pd

# Example data matrix
data = {
    'ID': [1, 2, 3, 4],
    'Feature1': [10, 15, 20, 25],
    'Feature2': [0.5, 1.2, 0.8, 1.0]
}

df = pd.DataFrame(data)
print("Data Matrix:")
print(df)
```

```python
# Dissimilarity Matrix

from scipy.spatial.distance import pdist, squareform
import pandas as pd

# Example data matrix
data = {
    'ID': [1, 2, 3, 4],
    'Feature1': [10, 15, 20, 25],
    'Feature2': [0.5, 1.2, 0.8, 1.0]
}

df = pd.DataFrame(data)

# Calculate Euclidean distance matrix
dissimilarity_matrix = squareform(pdist(df[['Feature1', 'Feature2']], metric='euclidean'))
print("Dissimilarity Matrix:")
print(pd.DataFrame(dissimilarity_matrix, index=df['ID'], columns=df['ID']))
```

    Pada contoh di atas, fungsi pdist dari SciPy menghitung jarak Euclidean antar pasangan instance berdasarkan fitur yang dipilih. Fungsi squareform mengubah matriks jarak berbentuk condensed menjadi bentuk persegi.

    Data matrix menyediakan representasi terstruktur dari data mentah, sedangkan dissimilarity matrix menangkap ketidakmiripan antar pasangan instance, yang berguna dalam berbagai aplikasi data mining seperti clustering, klasifikasi, dan similarity search.

### 2.3.2. Ukuran Proximity untuk Atribut Nominal (Proximity measures for nominal attributes)

Ukuran proximity untuk atribut nominal dipakai untuk mengukur kemiripan atau ketidakmiripan antar instance ketika berhadapan dengan atribut kategorik atau nominal. Berbeda dengan atribut numerik, atribut nominal tidak memiliki urutan alami sehingga metrik jarak biasa tidak dapat diterapkan. Ukuran proximity mengatasi persoalan ini dengan berfokus pada kesesuaian atau ketidaksesuaian antar nilai kategorik.

#### Contoh Ukuran Proximity: Jaccard Similarity

#### Penjelasan:

Jaccard Similarity mengukur kemiripan antara dua himpunan dengan membagi ukuran irisannya dengan ukuran gabungannya. Dalam konteks atribut nominal, setiap nilai atribut diperlakukan sebagai himpunan nilai biner, yaitu 1 jika atribut tersebut ada dan 0 jika tidak.

```python
from sklearn.metrics import jaccard_score

# Example dataset with nominal attributes
data = {
    'Instance1': ['Red', 'Square', 'Large'],
    'Instance2': ['Blue', 'Circle', 'Small']
}

# Convert nominal attributes to binary values
binary_data = pd.get_dummies(pd.DataFrame(data))
binary_data
```

```python
# Calculate Jaccard Similarity
jaccard_similarity = jaccard_score(binary_data.iloc[0], binary_data.iloc[1])
jaccard_similarity_1 = jaccard_score(binary_data.iloc[0], binary_data.iloc[0])
print(f"Jaccard Similarity: {jaccard_similarity}")
print(f"Jaccard Similarity: {jaccard_similarity_1}")
```

    Pada contoh di atas, atribut nominal ('Red', 'Square', 'Large') dan ('Blue', 'Circle', 'Small') diubah menjadi vektor biner menggunakan one-hot encoding. Jaccard Similarity kemudian dihitung menggunakan fungsi jaccard_score dari scikit-learn.

    Ukuran proximity untuk atribut nominal sangat penting dalam clustering, klasifikasi, dan tugas data mining lain yang melibatkan data kategorik. Contoh di atas memperlihatkan penerapan Jaccard Similarity untuk mengukur kemiripan antar instance dengan atribut nominal.

### 2.3.3. Ukuran Proximity untuk Atribut Biner (Proximity measures for binary attributes)

Ukuran proximity untuk atribut biner dipakai ketika berhadapan dengan dataset yang tersusun atas atribut biner bernilai 0 atau 1. Ukuran ini menilai kemiripan atau ketidakmiripan antar instance berdasarkan ada atau tidaknya fitur biner tertentu.

#### Contoh Ukuran Proximity: Hamming Distance

#### Penjelasan:

Hamming Distance mengukur ketidakmiripan antara dua string biner dengan menghitung banyaknya posisi yang bit-nya berbeda.

```python
import numpy as np
from scipy.spatial.distance import hamming

# Example dataset with binary attributes
data = {
    'Instance1': [1, 0, 1, 0],
    'Instance2': [0, 1, 1, 0]
}

# Convert binary attributes to NumPy arrays
binary_data = np.array([data['Instance1'], data['Instance2']])

# Calculate Hamming Distance
hamming_distance = hamming(binary_data[0], binary_data[1])
print(f"Hamming Distance: {hamming_distance}")
```

    Pada contoh di atas, instance direpresentasikan oleh vektor biner [1, 0, 1, 0] dan [0, 1, 1, 0]. Hamming Distance antara kedua vektor tersebut dihitung menggunakan fungsi hamming dari SciPy.

    Ukuran proximity untuk atribut biner sangat penting pada tugas seperti clustering, pengenalan pola, dan klasifikasi ketika dataset tersusun atas fitur biner. Contoh Hamming Distance di atas memperlihatkan cara mengukur ketidakmiripan antar instance berdasarkan atribut biner.

### 2.3.4. Dissimilarity Data Numerik: Minkowski Distance (Dissimilarity of numeric data)

Minkowski distance adalah ukuran dissimilarity yang dipakai untuk menilai kemiripan atau ketidakmiripan antara dua titik dalam ruang berdimensi banyak. Ukuran ini merupakan generalisasi dari beberapa ukuran jarak lain, termasuk Euclidean distance dan Manhattan distance, dan didefinisikan dengan rumus berikut:

$$D(x,y) = \left( \sum_{i=1}^{n} |x_i - y_i|^p \right)^{1/p}$$

dengan $x$ dan $y$ adalah vektor yang merepresentasikan titik data numerik, $n$ adalah banyaknya dimensi, dan $p$ adalah parameter yang menentukan orde Minkowski distance. Ketika $p=1$, Minkowski distance setara dengan Manhattan distance, dan ketika $p=2$, ukuran ini setara dengan Euclidean distance.

```python
from scipy.spatial.distance import minkowski

# Example dataset with numeric attributes
data1 = [2, 3, 5, 7]
data2 = [1, 4, 6, 8]

# Calculate Minkowski distance with p=2 (Euclidean distance)
minkowski_distance = minkowski(data1, data2, p=2)
print(f"Minkowski Distance (p=2): {minkowski_distance}")

# Calculate Minkowski distance with p=1 (Manhattan distance)
manhattan_distance = minkowski(data1, data2, p=1)
print(f"Minkowski Distance (p=1): {manhattan_distance}")
```

    Pada contoh di atas, Minkowski distance dihitung antara dua vektor numerik [2, 3, 5, 7] dan [1, 4, 6, 8] menggunakan Euclidean distance (p=2) dan Manhattan distance (p=1). Fungsi minkowski dari SciPy dipakai untuk perhitungannya.

    Minkowski distance adalah ukuran yang lentur dan menyesuaikan diri dengan berbagai keadaan bergantung pada nilai p yang dipilih. Ukuran ini berlaku pada berbagai tugas data mining seperti clustering, klasifikasi, dan deteksi outlier, terutama pada dataset yang berisi atribut numerik.

### 2.3.5. Ukuran Proximity untuk Atribut Ordinal (Proximity measures for ordinal attributes)

Ukuran proximity untuk atribut ordinal dipakai ketika berhadapan dengan dataset yang memuat atribut dengan urutan atau peringkat yang melekat. Berbeda dengan atribut nominal, atribut ordinal memiliki urutan yang bermakna, tetapi jarak antar nilainya belum tentu seragam.

#### Contoh Ukuran Proximity: Spearman Rank Correlation Coefficient

#### Penjelasan:

Spearman Rank Correlation Coefficient mengukur kekuatan dan arah hubungan monotonik antara dua variabel ordinal. Ukuran ini didasarkan pada peringkat nilainya, bukan pada nilai aslinya.

```python
import numpy as np
from scipy.stats import spearmanr

# Example dataset with ordinal attributes
data = {
    'Instance1': [2, 3, 1, 4],
    'Instance2': [1, 4, 2, 3]
}

# Calculate Spearman Rank Correlation Coefficient
spearman_corr, _ = spearmanr(data['Instance1'], data['Instance2'])
print(f"Spearman Rank Correlation Coefficient: {spearman_corr}")
```

    Pada contoh di atas, instance direpresentasikan oleh atribut ordinal [2, 3, 1, 4] dan [1, 4, 2, 3]. Spearman Rank Correlation Coefficient dihitung menggunakan fungsi spearmanr dari SciPy.

    Ukuran proximity untuk atribut ordinal penting ketika berhadapan dengan data yang memiliki urutan bermakna namun tidak memiliki skala numerik yang jelas. Spearman Rank Correlation Coefficient sangat berguna untuk menilai hubungan monotonik antar variabel ordinal.

### 2.3.6. Dissimilarity untuk Atribut Bertipe Campuran (Dissimilarity for attributes of mixed types)

Menangani dissimilarity untuk atribut bertipe campuran adalah tantangan yang umum ditemui dalam data mining, sebab dataset sering kali memuat gabungan atribut numerik, kategorik, dan ordinal sekaligus. Diperlukan ukuran dissimilarity yang sesuai dan mampu mengakomodasi keragaman tipe atribut tersebut.

#### Contoh Ukuran Dissimilarity: Gower's Distance

#### Penjelasan:

Gower's Distance adalah ukuran dissimilarity yang dirancang untuk dataset dengan tipe atribut campuran. Ukuran ini menghitung dissimilarity antara dua instance dengan mempertimbangkan tipe setiap atribut lalu menerapkan ukuran yang sesuai untuk masing-masing tipe.

Untuk atribut numerik, dissimilarity dihitung sebagai selisih absolut dibagi rentang atribut tersebut, sehingga hasilnya selalu berada pada rentang 0 sampai 1. Untuk atribut nominal dan biner, dissimilarity bernilai 0 jika kedua nilainya sama dan 1 jika berbeda. Untuk atribut ordinal, nilainya dinormalisasi lebih dahulu lalu diperlakukan seperti atribut numerik. Dissimilarity akhir adalah rata-rata dari seluruh atribut.

Karena setiap atribut sudah dinormalisasi ke rentang yang sama sebelum dirata-ratakan, tidak ada atribut yang mendominasi hasil hanya karena skalanya besar. Inilah yang membedakan Gower's Distance dari sekadar menghitung Euclidean distance atas gabungan atribut yang sudah di-encode.

```python
import numpy as np
import pandas as pd

# Example dataset with mixed attribute types
data = {
    'Numeric': [2.5, 1.8, 3.2, 4.0],
    'Binary': [1, 0, 1, 1],
    'Nominal': ['A', 'B', 'A', 'C'],
    'Ordinal': [2, 1, 3, 2]
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
print("\n")


def gower_distance(data, numeric_columns, categorical_columns):
    """Calculate the Gower's Distance matrix for mixed attribute types.

    Numeric and ordinal attributes: absolute difference divided by the range.
    Binary and nominal attributes: 0 if the values match, 1 otherwise.
    The final distance is the average over all attributes.
    """
    n_rows = len(data)
    total = np.zeros((n_rows, n_rows))

    # Numeric and ordinal attributes are scaled by their own range
    for column in numeric_columns:
        values = data[column].to_numpy(dtype=float)
        value_range = values.max() - values.min()
        value_range = value_range if value_range > 0 else 1.0
        total += np.abs(values[:, None] - values[None, :]) / value_range

    # Binary and nominal attributes use the simple matching coefficient
    for column in categorical_columns:
        values = data[column].to_numpy()
        total += (values[:, None] != values[None, :]).astype(float)

    return total / (len(numeric_columns) + len(categorical_columns))


# Ordinal attributes are treated as numeric after normalization
numeric_attributes = ['Numeric', 'Ordinal']
categorical_attributes = ['Binary', 'Nominal']

# Calculate Gower's Distance
gower_matrix = gower_distance(df, numeric_attributes, categorical_attributes)

print("Gower's Distance Matrix:")
print(np.round(gower_matrix, 3))
```

    Pada contoh di atas, Gower's Distance dihitung untuk dataset dengan atribut numerik, biner, nominal, dan ordinal sekaligus. Setiap atribut dinormalisasi ke rentang 0 sampai 1 lebih dahulu, lalu seluruh hasilnya dirata-ratakan.

    Karena hasilnya berupa rata-rata dari nilai yang sudah dinormalisasi, seluruh sel pada matriks berada di antara 0 dan 1. Nilai 0 berarti kedua instance identik pada seluruh atribut, sedangkan nilai 1 berarti keduanya berbeda sejauh mungkin. Diagonal matriks selalu bernilai 0 karena jarak sebuah instance terhadap dirinya sendiri adalah nol.

    Gower's Distance adalah ukuran yang serbaguna untuk dissimilarity pada dataset dengan tipe atribut campuran, dan menyediakan solusi menyeluruh untuk menangani berbagai karakteristik data.

### 2.3.7. Cosine Similarity

Cosine similarity adalah metrik yang dipakai untuk mengukur kemiripan antara dua vektor tak nol pada suatu ruang hasil kali dalam. Dalam konteks data mining, metrik ini sering dipakai untuk menilai kemiripan antar dokumen dalam natural language processing, atau untuk membandingkan vektor fitur dalam sistem rekomendasi. Nilai cosine similarity berada pada rentang -1 sampai 1, dengan 1 menandakan kemiripan sempurna, 0 menandakan tidak ada kemiripan, dan -1 menandakan ketidakmiripan sempurna.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Example dataset with text documents
documents = [
    "Data mining is an exciting field of study.",
    "Machine learning techniques are used for data analysis.",
    "Python programming is widely used in data science applications."
]

# Convert documents to a bag-of-words representation
vectorizer = CountVectorizer()
document_matrix = vectorizer.fit_transform(documents)

# Calculate cosine similarity
cosine_similarities = cosine_similarity(document_matrix, document_matrix)
print("Cosine Similarity Matrix:")
print(cosine_similarities)
```

    Pada contoh di atas, CountVectorizer dari scikit-learn dipakai untuk mengubah kumpulan dokumen teks menjadi matriks jumlah kemunculan kata (representasi bag-of-words). Fungsi cosine_similarity kemudian menghitung cosine similarity antar vektor dokumen.

    Cosine similarity sangat berguna pada situasi ketika besar vektor tidak penting dan yang menjadi fokus adalah arah vektornya. Metrik ini banyak dipakai dalam analisis teks, temu kembali dokumen, dan sistem collaborative filtering.

### 2.3.8. Mengukur Kemiripan Distribusi: Kullback-Leibler Divergence (Measuring similar distributions)

Kullback-Leibler (KL) Divergence adalah ukuran seberapa jauh sebuah distribusi peluang menyimpang dari distribusi peluang lain yang menjadi acuan. Dalam konteks data mining, ukuran ini sering dipakai untuk mengukur perbedaan atau kemiripan antara dua distribusi peluang. KL Divergence bukan metrik jarak sejati karena tidak simetris dan tidak memenuhi pertidaksamaan segitiga.

```python
import numpy as np
from scipy.stats import entropy

# Example probability distributions
distribution1 = np.array([0.2, 0.3, 0.5])
distribution2 = np.array([0.1, 0.6, 0.3])

# Calculate Kullback-Leibler Divergence
kl_divergence = entropy(distribution1, distribution2)
print(f"Kullback-Leibler Divergence: {kl_divergence}")
```

    Pada contoh di atas, dua distribusi peluang direpresentasikan oleh array distribution1 dan distribution2. Fungsi entropy dari SciPy menghitung KL Divergence antara kedua distribusi tersebut.

    KL Divergence sering dipakai dalam teori informasi dan statistik untuk mengukur perbedaan antara dua distribusi peluang. Ukuran ini diterapkan pada berbagai bidang, termasuk natural language processing, machine learning, dan pengenalan pola.

### 2.3.9. Menangkap Semantik Tersembunyi dalam Ukuran Kemiripan (Capturing hidden semantics in similarity measures)

Menangkap semantik tersembunyi dalam ukuran kemiripan berarti menemukan pola, hubungan, atau kemiripan yang bermakna dalam data namun tidak tampak pada pandangan pertama. Proses ini biasanya memerlukan teknik yang lebih maju, melampaui metrik jarak atau kemiripan yang sederhana. Metode seperti word embeddings, semantic similarity, atau model neural network tingkat lanjut dipakai untuk menangkap dan memanfaatkan semantik tersembunyi dalam data.

```python
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

# Example dataset with text documents
documents = [
    "Data mining is an exciting field of study.",
    "Machine learning techniques are used for data analysis.",
    "Python programming is widely used in data science applications."
]

# Tokenize the documents into words
tokenized_documents = [doc.split() for doc in documents]

# Train a Word2Vec model on the tokenized documents
model = Word2Vec(tokenized_documents, vector_size=10, window=3, min_count=1, workers=4)

# Calculate cosine similarity using word embeddings
embedding_similarity = cosine_similarity([model.wv['data']], [model.wv['mining']])
print(f"Cosine Similarity using Word Embeddings: {embedding_similarity[0][0]}")
```

    Pada contoh di atas, library Gensim dipakai untuk melatih model Word2Vec pada dataset kecil berisi dokumen teks. Word embeddings menangkap hubungan semantik antar kata, lalu cosine similarity dipakai untuk mengukur kemiripan antara vektor kata 'data' dan 'mining'.

    Menangkap semantik tersembunyi umumnya melibatkan teknik lanjutan seperti word embeddings, neural network, atau model deep learning. Metode-metode ini sangat berguna pada aplikasi seperti natural language processing, sistem rekomendasi, dan analisis citra, yang menuntut penangkapan makna atau semantik yang mendasari datanya.
